```python
import cv2
```

```python
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('sachin.jpg',0)
plt.imshow(img,cmap='gray')
```

```python
plt.imshow(img,cmap='Blues')
```

```python
# Reversing the image


orig_map = plt.colormaps['Blues']
reversed_map = orig_map.reversed()
plt.imshow(img, cmap = reversed_map)
```

```python
orig_map = plt.colormaps['Blues'].reversed()
```

```python
orig_map
```

```python
plt.figure(figsize=(16,8))
plt.subplot(2,4,1);  plt.title('Blues')
plt.imshow(img, cmap = plt.colormaps['Blues'].reversed())
plt.subplot(2,4,2);  plt.title('Greens')
plt.imshow(img, cmap = plt.colormaps['Greens'].reversed())
plt.subplot(2,4,3);  plt.title('Reds')
plt.imshow(img, cmap = plt.colormaps['Reds'].reversed())
plt.subplot(2,4,4);  plt.title('Oranges')
plt.imshow(img, cmap = plt.colormaps['Oranges'].reversed())
plt.subplot(2,4,5);  plt.title('Purples')
plt.imshow(img, cmap = plt.colormaps['Purples'].reversed())
plt.subplot(2,4,6);  plt.title('Greys')
plt.imshow(img, cmap = plt.colormaps['Greys'].reversed())
plt.subplot(2,4,7);  plt.title('YlOrBr')
plt.imshow(img, cmap = plt.colormaps['YlOrBr'].reversed())
plt.subplot(2,4,8);  plt.title('OrRd')
plt.imshow(img, cmap = plt.colormaps['OrRd'].reversed())
```

```python
plt.figure(figsize=(16,8))
plt.subplot(2,4,1);  plt.title('YlOrBr')
plt.imshow(img, cmap = plt.colormaps['YlOrBr'].reversed())
plt.subplot(2,4,2);  plt.title('OrRd')
plt.imshow(img, cmap = plt.colormaps['OrRd'].reversed())
plt.subplot(2,4,3);  plt.title('PuRd')
plt.imshow(img, cmap = plt.colormaps['PuRd'].reversed())
plt.subplot(2,4,4);  plt.title('BuPu')
plt.imshow(img, cmap = plt.colormaps['BuPu'].reversed())
plt.subplot(2,4,5);  plt.title('GnBu')
plt.imshow(img, cmap = plt.colormaps['GnBu'].reversed())
plt.subplot(2,4,6);  plt.title('Greys')
plt.imshow(img, cmap = plt.colormaps['PuBu'].reversed())
plt.subplot(2,4,7);  plt.title('PuBu')
plt.imshow(img, cmap = plt.colormaps['PuBu'].reversed())
plt.subplot(2,4,8);  plt.title('YlGn')
plt.imshow(img, cmap = plt.colormaps['YlGn'].reversed())
```

```python
#diverging
plt.figure(figsize=(16,8))
plt.subplot(2,4,1);  plt.title('PiYG')
plt.imshow(img, cmap = plt.colormaps['PiYG'].reversed())
plt.subplot(2,4,2);  plt.title('PRGn')
plt.imshow(img, cmap = plt.colormaps['PRGn'].reversed())
plt.subplot(2,4,3);  plt.title('BrBG')
plt.imshow(img, cmap = plt.colormaps['BrBG'].reversed())
plt.subplot(2,4,4);  plt.title('PuOr')
plt.imshow(img, cmap = plt.colormaps['PuOr'].reversed())
plt.subplot(2,4,5);  plt.title('RdGy')
plt.imshow(img, cmap = plt.colormaps['RdGy'].reversed())
plt.subplot(2,4,6);  plt.title('RdBu')
plt.imshow(img, cmap = plt.colormaps['RdBu'].reversed())
plt.subplot(2,4,7);  plt.title('coolwarm')
plt.imshow(img, cmap = plt.colormaps['coolwarm'].reversed())
plt.subplot(2,4,8);  plt.title('seismic')
plt.imshow(img, cmap = plt.colormaps['seismic'].reversed())
```

```python

#cyclic
plt.figure(figsize=(16,8))
plt.subplot(2,4,1);  plt.title('twilight')
plt.imshow(img, cmap = plt.colormaps['twilight'].reversed())
plt.subplot(2,4,2);  plt.title('twilight_shifted')
plt.imshow(img, cmap = plt.colormaps['twilight_shifted'].reversed())
plt.subplot(2,4,3);  plt.title('hsv')
plt.imshow(img, cmap = plt.colormaps['hsv'].reversed())
plt.subplot(2,4,4);  plt.title('viridis')
plt.imshow(img, cmap = plt.colormaps['viridis'].reversed())
plt.subplot(2,4,5);  plt.title('plasma')
plt.imshow(img, cmap = plt.colormaps['plasma'].reversed())
plt.subplot(2,4,6);  plt.title('inferno')
plt.imshow(img, cmap = plt.colormaps['inferno'].reversed())
plt.subplot(2,4,7);  plt.title('twilight')
plt.imshow(img, cmap = plt.colormaps['twilight'].reversed())
plt.subplot(2,4,8);  plt.title('twilight_shifted')
plt.imshow(img, cmap = plt.colormaps['twilight_shifted'].reversed())

plt.imshow(img, cmap = plt.colormaps['seismic'].reversed())
```
