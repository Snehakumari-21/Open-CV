```python
import numpy as np
```

```python
#Creating chess board

x = np.zeros([64], dtype=np.uint8)

x = x.reshape(8,8)
#creating chessboard
for i in range(8):
    for j in range(8):
        if( (i+j)%2 != 0):
            x[i,j] = 0
        else:
            x[i,j] = 1
plt.imshow(x,cmap='gray')
```
